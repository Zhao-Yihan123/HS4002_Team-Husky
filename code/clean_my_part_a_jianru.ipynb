# Cleaning Dependent Variable Subjective Security Q131

```python
# Selected data Q131
my['Q131'].value_counts()
```

```python
# Check NA
my['Q131'].isna().sum()
# No NA in my['Q131']
```

```python
# Recode Q131 into an ordinal variable (categorical) and rename it into Q131_cat.
my['Q131_cat'] = pd.Categorical(
    my['Q131'],
    categories=[1, 2, 3, 4],
    ordered=True
)
my['Q131_cat']
# 1 is very safe, 4 is very unsafe.
```

```python
# check cleaned data
my['Q131_cat'].unique()
my['Q131_cat'].value_counts()
my['Q131_cat'].value_counts().sort_index()
```

# Cleaning Independent Variable perceived prevalence of sexual harassment Q138

```python
# Selected data Q138
my['Q138'].value_counts()
```

```python
# Check NA
my['Q138'].isna().sum()
# No NA in my['Q138']
```

```python
# Recode Q138 into an ordinal variable (categorical) and rename it into Q138_cat.
my['Q138_cat'] = pd.Categorical(
    my['Q138'],
    categories=[1, 2, 3, 4],
    ordered=True
)
my['Q138_cat']
# 1 is very frequently, 4 is never.
```

```python
# check cleaned data
my['Q138_cat'].unique()
my['Q138_cat'].value_counts()
my['Q138_cat'].value_counts().sort_index()
```

```python
# IMPORTANT:
# Q131: higher values indicate lower self-perceived security.
# BUT for Q138: higher values indicate lower self-perceived prevalence of sexual harassment.
```
