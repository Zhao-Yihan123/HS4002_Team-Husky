```python
# Malaysia Part B - Data Cleaning
# Sutong

import pandas as pd
import numpy as np
```

```python
my = pd.read_csv("../data/raw/wvs7_malaysia.csv", sep=";", 
index_col=False) # Prevent pandas from mistakenly treating 1st data column as the index


print(my.shape) #（rows, columns)
my[["Q273", "Q275", "Q279", "Q288", "Q144", "Q145"]].head() #education (Q275); household income (Q288); employment status (Q279); marital status (Q273); personal or family crime victimisation (Q144–Q145)
```

```python
vars_b = ["Q275", "Q288", "Q279", "Q273", "Q144", "Q145"]

for var in vars_b:
    print(var)
    print(my[var].value_counts(dropna=False).sort_index())
```

```python
#define missing value
wvs_missing = [-1, -2, -3, -5]
```

## Clean Education Level
0 = Early childhood / no education
1 = Primary
2 = Lower secondary
3 = Upper secondary
4 = Post-secondary non-tertiary
5 = Short-cycle tertiary
6 = Bachelor
7 = Master
8 = Doctoral

```python
#check distribution of score
my["Q275"].value_counts(dropna=False).sort_index()
```

```python
#Although it seems like there is no missing value, just replace missing value for na out of precaution
my["education"] = my["Q275"].replace(wvs_missing, np.nan)
```

```python
#rename education column to edu
my = my.rename(columns={"education": "edu"})
```

```python
#check
my["edu"].value_counts(dropna=False).sort_index()

my.loc[
    my["edu"].notna() & ~my["edu"].between(0, 8),
]
```

## Clean Percieved Household Income Rank

```python
# Check the data type of household income rank
my["Q288"].dtype
```

```python
my["household_income"] = my["Q288"].replace(wvs_missing, np.nan)
my["household_income"].value_counts(dropna=False).sort_index()
```

```python
# Check any abnormal value: non-missing values but outside the valid 1-10 range
# ~ means not
my.loc[
    my["household_income"].notna() & ~my["household_income"].between(1, 10),
    ["Q288", "household_income"] #see original data and
] #loc[row wanted, column wanted]
```

## Clean Employment Status (Q279)
- **1** = Full-time employee (30 hours a week or more)
- **2** = Part-time employee (less than 30 hours a week)
- **3** = Self-employed
- **4** = Retired / pensioned
- **5** = Housewife not otherwise employed
- **6** = Student
- **7** = Unemployed
- **8** = Other

```python
#see distribution
my["Q279"].value_counts(dropna=False).sort_index()

#from the output, we can see there is no missing value
```

```python
my["em"] = my["Q279"].map(
    {1:1,
    2:1, 
    3:1,
    4:0, 
    5:0, 
    6:0, 
    7:0, 
    8:0}
)
```

```python
#check
my["em"].value_counts(dropna=False).sort_index
```

```python
#check
my[["Q279", "em"]].head(20)
```

## Clean marital status (Q273)
- 1 Married
- 2 Living together as married
- 3 Divorced
- 4 Separated
- 5 Widowed
- 6 Single

```python
# see distribution
my["Q273"].value_counts(dropna=False)

#there is no missing value
```

```python
#recode it in a binary variable: 1 = married; 0 = unmarried
my["marriage"] = my["Q273"].map(
    {
        1:1,
        2:1,
        3:0,
        4:0,
        5:0,
        6:0
    }
)
```

```python
#check
my["marriage"].value_counts(dropna=False)
```

```python
#check
my[["Q273", "marriage"]].head(20)
```

## Clean personal or family crime victimisation (Q144–Q145)
1 = Yes
2 = No

```python
my["Q144"].value_counts(dropna=False).sort_index()
#output shows that there is no missing value
```

```python
my['victim_self']=my["Q144"]
```

```python
# Recode Q144 into a binary variable: 1 = victimised, 0 = not victimised
my["victim_self"] = my["Q144"].map({
    1: 1,
    2: 0
})
```

```python
#check
my[["Q144", "victim_self"]].head(20)
```

```python
my["Q145"].value_counts(dropna=False).sort_index()
#output shows that there is no missing value
```

```python
my["victim_family"] = my["Q145"].map(
    {1: 1,
    2: 0}
)
```

```python
my[["Q145", "victim_family"]].head(20)
```
